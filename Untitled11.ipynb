```python
num=[10,40,30,15,17,60]

for i in range(len(num)):
    for j in range(i+1,len(num)):
        if num[i] > num[j]:
            num[i],num[j] = num[j],num[i]

print(num)
```

```python
num=[10,40,30,15,17,60]
new=[]

while num:
    small=num[0]

    for i in range(len(num)):
        if num[i] < small:
            small=num[i]

    new.append(small)
    num.remove(small)

print(new)
```

```python
num=[10,7,6,14,9,13]

buy=num[0]
profit=0

for i in range(len(num)):
    if num[i] < buy:
        buy=num[i]

    if num[i]-buy > profit:
        profit=num[i]-buy

print(profit)
```
