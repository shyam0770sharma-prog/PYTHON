```python
file=open('/california_housing_shyam_small.csv')
content = file.read()

file.close()

#print(content)
```

```python
import csv

data_dict = {}

for index, row in enumerate(csv.reader(content.strip().split('\n'))):
    if index == 0:
        # Initialize dictionary with column headers as keys
        for column in row:
            data_dict[column] = []
    else:
        # Append each element in the row to the correct list in the dictionary
        for col_index, value in enumerate(row):
            data_dict[__builtins__.list(data_dict.keys())[col_index]].append(value)
```
