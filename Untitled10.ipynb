```python
num = int(input("Enter num"))
operator = input("Enter operator (+ - * /): ")
num2 = int(input("Enter num2"))

if operator == '+':
    result = num + num2
    print(result)
elif operator == '-':
    result = num - num2
    print(result)
elif operator == '*':
    result = num * num2
    print(result)
elif operator == '/' :
        result = num / num2
        print(result)
else:
    print("Enter valid input")
```

```python
weight = float(input("Enter your weight"))
height = float(input("Enter your height"))

BMI= weight/(height**2)
if(BMI < 18.5):
    print("underweight")
elif(18.5 <= BMI < 24.9):
    print("normal weight")
elif(25 <= BMI <29.9):
    print("overweight")
else:
    print("obesity")
```
