```python
var = "saras"
index = 0
end = len(var)-1
a = False
while index < len(var):
    if var[index] != var[end]:
        a = True
        break

    index += 1
    end -= 1

if(a==True):
    print("not a palindrom")
else:
    print("palindrom")
```

```python
#Q1. Find the second-highest salary
salaries = [50000, 70000, 45000, 90000, 70000, 80000]
HighesSalary=0
SecondHighest=0
for i in range(0,len(salaries)):
  if salaries[i] > HighesSalary:
    SecondHighest=HighesSalary
    HighesSalary=salaries[i]
  elif salaries[i]>SecondHighest and salaries[i] != HighesSalary:
      SecondHighest=salaries[i]
print(SecondHighest)
```

```python
#Q2. Find duplicate values
numbers = [10, 20, 10, 30, 40, 20, 50, 30]
original=[]
duplicate=[]
for i in numbers:
 if i in original:
   duplicate.append(i)
 else:
    original.append(i)
print(duplicate)
```

```python
#Q4. Find the maximum profit
prices = [10, 7, 6, 14, 8, 13]

left = 0
right = 1

max_profit = 0

while right < len(prices):

    if prices[left] < prices[right]:

        profit = prices[right] - prices[left]

        if profit > max_profit:
            max_profit = profit

    else:
        left = right

    right += 1

print(max_profit)
```
