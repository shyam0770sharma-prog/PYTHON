```python
df_india = df[df['job_location'] == 'India']
```

```python
df_india.reset_index(inplace=True)

df_india.head(5)
```

```python
median_pivot = df_india.pivot_table(values= 'salary_year_avg', index='job_title_short', aggfunc=['median','min','max'])
median_pivot
```

```python
median_pivot.sort_values(by=('median','salary_year_avg'), ascending=False)
```
