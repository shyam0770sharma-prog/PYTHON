```python
#6. Write a program to count the number of digits in a number.
num = 12345
count = 0

while num > 0:
    num = num // 10
    count += 1
print(count)
```

```python
#7. Write a program to calculate the sum of digits of a number.
sum = 0
num = 12345

while num > 0:
    digit = num % 10
    sum += digit
    num = num // 10

print(sum)
```

```python
#8. Write a program to find the first and last digit of a number.
num = 12345

last_digit = num % 10

while num > 9:
  num = num // 10


  first_digit = num

print(first_digit, last_digit)
```

```python
#10.Write a program to enter a number and print its reverse.
num=12345
rev=0

while num>0:
    digit=num%10
    rev=rev*10+digit
    num//=10

print(rev)
```
